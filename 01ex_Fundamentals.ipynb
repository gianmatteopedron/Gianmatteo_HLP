4\. Write the factorial function so that it a) does and b) does not use recursion.

````python
def fact (x):

    if x ==0 or x == 1:
        return 1
    
    return x*fact(x-1)

x = 5
print (fact(x))

def fact_norec(y):

    if y == 0:
        return 1

    fact = 1
    for i in range(1, y+1):
        fact = fact * i
    return fact


y = 5
print (fact_norec(y))
    
````
